# Model Training on Processed Data
This notebook loads the processed training and testing sets from Google Drive and trains five different classification models.

```python
import pandas as pd
import joblib
import os
import shutil
from sklearn.metrics import classification_report, accuracy_score

# Path where processed data is stored
drive_path = '/content/drive/MyDrive/processed_data'

# Load datasets
train_df = pd.read_csv(os.path.join(drive_path, 'train_processed.csv'))
test_df = pd.read_csv(os.path.join(drive_path, 'test_processed.csv'))

# Separate features and target
X_train = train_df.drop('Churn', axis=1)
y_train = train_df['Churn']
X_test = test_df.drop('Churn', axis=1)
y_test = test_df['Churn']

print("Data loaded from Google Drive.")
```

### 1. Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

# Initialize and train
lr_model = LogisticRegression(max_iter=1000, random_state=42)
lr_model.fit(X_train, y_train)

# Evaluation
print(f"Logistic Regression Accuracy: {accuracy_score(y_test, lr_model.predict(X_test)):.4f}")

# Save to Drive
model_file = 'logistic_regression_model.pkl'
joblib.dump(lr_model, model_file)
shutil.move(model_file, os.path.join(drive_path, model_file))
print(f"Saved to {drive_path}")
```

### 2. Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier

# Initialize and train
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)

# Evaluation
print(f"Decision Tree Accuracy: {accuracy_score(y_test, dt_model.predict(X_test)):.4f}")

# Save to Drive
model_file = 'decision_tree_model.pkl'
joblib.dump(dt_model, model_file)
shutil.move(model_file, os.path.join(drive_path, model_file))
print(f"Saved to {drive_path}")
```

### 3. Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

# Initialize and train
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# Evaluation
print(f"Random Forest Accuracy: {accuracy_score(y_test, rf_model.predict(X_test)):.4f}")

# Save to Drive
model_file = 'random_forest_model.pkl'
joblib.dump(rf_model, model_file)
shutil.move(model_file, os.path.join(drive_path, model_file))
print(f"Saved to {drive_path}")
```

### 4. Support Vector Machine (SVM)

```python
from sklearn.svm import SVC

# Initialize and train
svm_model = SVC(probability=True, random_state=42)
svm_model.fit(X_train, y_train)

# Evaluation
print(f"SVM Accuracy: {accuracy_score(y_test, svm_model.predict(X_test)):.4f}")

# Save to Drive
model_file = 'svm_model.pkl'
joblib.dump(svm_model, model_file)
shutil.move(model_file, os.path.join(drive_path, model_file))
print(f"Saved to {drive_path}")
```

### 5. XGBoost

```python
from xgboost import XGBClassifier

# Initialize and train
xgb_model = XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42)
xgb_model.fit(X_train, y_train)

# Evaluation
print(f"XGBoost Accuracy: {accuracy_score(y_test, xgb_model.predict(X_test)):.4f}")

# Save to Drive
model_file = 'xgboost_model.pkl'
joblib.dump(xgb_model, model_file)
shutil.move(model_file, os.path.join(drive_path, model_file))
print(f"Saved to {drive_path}")
```
